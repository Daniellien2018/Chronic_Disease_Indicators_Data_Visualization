## Data Cleaning - Chronic Obstructive Pulmonary Disease (COPD)

```python
# Begin Disease Specified Cleaning
df = df_copy.copy()
df.shape #should give (309215, 34)
```

```python
df.isnull().sum() #Check for missing values from entire dataset
```

```python
# Drop Columns from entire dataset that are missing = Num of missing values == num of rows --> Completely empty
df = df.drop(columns=[
            'Response', 
            'StratificationCategory2',
            'Stratification2',
            'StratificationCategory3',
            'Stratification3',
            'ResponseID',
            'StratificationCategoryID2', 
            'StratificationID2',         
            'StratificationCategoryID3',
            'StratificationID3'         
        ])
print(df.shape)
#should be (1185676, 24) - 10 cols were dropped
```

```python
# Now filter entier dataset for topic = 'Chronic Obstructive Pulmonary Disease'
topic = 'Chronic Obstructive Pulmonary Disease'
df_copd = df[df['Topic'] == topic]
print(df_copd.shape)
# df_copd.head()
# (1185676, 24) to (152874, 24) - about 12.9% of data is related to COPD
```

```python
df_copd.isnull().sum()
```

```python
# Drop More columns that are not needed for COPD
columns_to_drop = [
    'YearEnd',
    'LocationAbbr',
    'DataSource',
    'DataValueAlt',
    'DataValueFootnoteSymbol',
    'DatavalueFootnote',
    'HighConfidenceLimit',
    'LowConfidenceLimit',
    'LocationID',
    'TopicID',
    'DataValueTypeID'
]

df_copd = df_copd.drop(columns=columns_to_drop)

# # From 34 columns to 13 columns - correct
df_copd.shape
```

```python
df_copd.isnull().sum()
```

```python
# Last Check
df_copd.head()
```

```python
# Reset Index
df_copd.reset_index(drop=True, inplace=True)

```

```python
# save to csv
output_dir = '/home/daniel.lien/dev/homework/data/cleaned_copd_data_2.csv'
df_copd.to_csv(output_dir, index=False)
```
