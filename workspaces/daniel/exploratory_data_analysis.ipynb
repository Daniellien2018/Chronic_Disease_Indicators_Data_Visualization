## Data Cleaning - Chronic Obstructive Pulmonary Disease (COPD)

```python
# Drop Columns that are not needed or empty
df = df.drop(columns=['Response','ResponseID',
              'Stratification2','StratificationCategory2','StratificationID2','StratificationCategoryID2',
              'Stratification3', 'StratificationCategory3', 'StratificationID3','StratificationCategoryID3'])

# From 34 columns to 24 columns - correct
df.head()
```

```python
df_copd = df[df['Topic'] == 'Chronic Obstructive Pulmonary Disease']
print(df_copd.shape)
# df_copd.head()
# #(309215, 34) to (26951, 29) - about 8.7% of data is related to COPD
```

```python
df_copd.isnull().sum()
```

```python
#Drop more uncessary columns
"""
Data Value Alt (copy of data value)
Data Value Footnote 
Data Value Footnote Symbol (too many missing values)
Location Abbr (redundant) We are keeping location Desc
Location ID (redundant)
TopicID (we should only have COPD) --> confirmed print(df_copd["TopicID"].unique())
Data Value Type ID (we will use DataValueType)
"""
columns_to_drop = ['DataValueAlt', 'DataValueFootnoteSymbol', 'DataValueFootnote',
                   'LocationID', 'LocationAbbr', 'TopicID', 'DataValueTypeID']

df_copd = df_copd.drop(columns=columns_to_drop)
```

```python
df_copd.isnull().sum()
```

```python
# Last Check
df_copd.shape
df_copd.head()
```

```python
# Reset Index
df_copd.reset_index(drop=True, inplace=True)
```

```python
#create a working copy
df = df_copd.copy()
print(df.shape)
df.head()
```
