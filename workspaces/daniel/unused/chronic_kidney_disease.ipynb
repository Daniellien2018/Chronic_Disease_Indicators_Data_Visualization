```python
import pandas as pd
path = '/home/daniel.lien/dev/homework/data/U.S._Chronic_Disease_Indicators__CDI___2023_Release copy.csv'
def read_data(file_path):
    return pd.read_csv(file_path)
df_all = read_data(path)
df_all.head()
```

```python
df = df_all[df_all['Topic'] == 'Chronic Kidney Disease']
```

```python
columns_to_drop = ['Response', 
            'StratificationCategory2',
            'Stratification2',
            'StratificationCategory3',
            'Stratification3',
            'ResponseID',
            'StratificationCategoryID2', 
            'StratificationID2',         
            'StratificationCategoryID3',
            'StratificationID3',
            'YearEnd',
            'LocationAbbr',
            'DataSource',
            'DataValueAlt',
            'DataValueFootnoteSymbol',
            'HighConfidenceLimit',
            'LowConfidenceLimit',
            'LocationID',
            'TopicID',
            'DataValueTypeID'
    ]
df = df.drop(columns=columns_to_drop)
```

```python
df.head()
```

```python
# Group by Question and Question ID to Create Question Map
question_ID_map = df.groupby("Question")['QuestionID'].unique().apply(sorted).to_dict()
# pd.DataFrame([(k,v) for k, v in question_ID_map.items()], columns=['Question', 'QuestionID'])
question_df = (pd.DataFrame([(k, v) for k, ids in question_ID_map.items() for v in ids], 
                             columns=['Question', 'QuestionID'])
               .sort_values(by='QuestionID')
               .reset_index(drop=True))

# Print the DataFrame nicely
print(question_df)
```
