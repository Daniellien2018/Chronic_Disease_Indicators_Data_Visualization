## Given the Starting Dataset, curate the dataset that applied to COPD


```python
import pandas as pd
from pathlib import Path
```

```python

rel_path = Path("/home/daniel.lien/dev/homework/data/")
dataset_csv = Path("U.S._Chronic_Disease_Indicators__CDI___2023_Release copy.csv") # Start with Main Dataset 

```

```python
# Step 1: Read in Data
df = pd.read_csv(rel_path / dataset_csv)
print(df.shape)
```

```python
# Step 2: Filter The Data for COPD-related records
df = df[df['Topic'] == 'Chronic Obstructive Pulmonary Disease']
print(df.shape)
```

```python
# Percentage of Rows that are COPD prelated
copd_percent = (152874 / 1185676) * 100
print(copd_percent) 
```

```python
df.columns
```

```python
#Drop Cols
columns_to_drop = ['Response', 
        'StratificationCategory2',
        'Stratification2',
        'StratificationCategory3',
        'Stratification3',
        'ResponseID',
        'StratificationCategoryID2', 
        'StratificationID2',         
        'StratificationCategoryID3',
        'StratificationID3',
        'YearEnd',
        'LocationAbbr',
        'DataSource',
        'DataValueAlt',
        'DataValueFootnoteSymbol',
        'DatavalueFootnote',
        'HighConfidenceLimit',
        'LowConfidenceLimit',
        'LocationID',
        'TopicID',
        'DataValueTypeID',
        'GeoLocation'
]
df = df.drop(columns=columns_to_drop)
df.head()
```

```python
df = df.dropna(subset=['DataValue'])

# Remove non-state locations
states_to_keep = ["Alabama", "Alaska", "Arizona", "Arkansas", "California", "Colorado", "Connecticut", "Delaware", 
                  "Florida", "Georgia", "Hawaii", "Idaho", "Illinois", "Indiana", "Iowa", "Kansas", "Kentucky", 
                  "Louisiana", "Maine", "Maryland", "Massachusetts", "Michigan", "Minnesota", "Mississippi", "Missouri", 
                  "Montana", "Nebraska", "Nevada", "New Hampshire", "New Jersey", "New Mexico", "New York", "North Carolina", 
                  "North Dakota", "Ohio", "Oklahoma", "Oregon", "Pennsylvania", "Rhode Island", "South Carolina", "South Dakota", 
                  "Tennessee", "Texas", "Utah", "Vermont", "Virginia", "Washington", "West Virginia", "Wisconsin", "Wyoming"]

df = df[df['LocationDesc'].isin(states_to_keep)]
df.shape
```

```python
df.to_csv(rel_path / "new_copd_data.csv", index=False)
print("Done")
```
