```python
import pandas as pd
```

```python
medicaid = '/home/daniel.lien/dev/homework/data/medicaid_data_cleaned.csv'
medicaid_df = pd.read_csv(medicaid)
```

```python
median_income = '/home/daniel.lien/dev/homework/data/medianincome_temp.csv'
median_income_df = pd.read_csv(median_income)   
```

```python
copd = '/home/daniel.lien/dev/homework/data/cleaned_copd_data_final.csv'
copd_df = pd.read_csv(copd)
```

```python
"""
Median Income: State, Year, Median_Income
Medicare Spending: State, Year, Medicaid_Spending
COPD: Year, State, Question, Mortality, Stratification
"""
```

```python
# Merge median income into COPD data
df_merged = copd_df.merge(median_income_df, on=['State', 'Year'], how='left')

# Merge Medicare spending into the result
df_merged = df_merged.merge(medicaid_df, on=['State', 'Year'], how='left')
```

```python
# Drop question column
df_merged = df_merged.drop(columns='Question')
df_merged.head()
```

```python
output_dir = '/home/daniel.lien/dev/homework/data/final_df.csv'
df_merged.to_csv(output_dir, index=False)
```
