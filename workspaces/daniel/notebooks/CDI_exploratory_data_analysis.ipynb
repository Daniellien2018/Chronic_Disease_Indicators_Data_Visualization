# Exploratory Data Analysis for Chronic Disease Indicator - All diseases

# Context
Chronic Disease Indicators provides data that tracks various chronic diseases and their risk factors at the national and state levels.
This dataset is a compliation of various data sources, including surveys, vital records, and administrative data. 

## Data Dictionary:
- 'YearStart': Year the Survey/Collection began
- 'YearEnd': Year the Survey/Collection ended
- 'LocationAbbr': Two (2) letter abbreviation of the Location
- 'LocationDesc': Full name of the Location
- 'DataSource': Survey/Collection Source of the data
- 'Topic': Chronic Disease Topic
- 'Question': Question related to the Topic
- 'Response': Nan
- 'DataValueUnit': Response Value unit type
- 'DataValueType'" Response Value type
- 'DataValue': Response Value
- 'DataValueAlt': Alternate Response Value
- 'DataValueFootnoteSymbol': Footnote symbol
- 'DatavalueFootnote': Footnote 
- 'LowConfidenceLimit': Lowerbound confidence level
- 'HighConfidenceLimit': Higherbound confidence level
- 'StratificationCategory1': Stratification category for response
- 'Stratification1': Stratification grouping
- 'StratificationCategory2', Secondary Stratification category
- 'Stratification2', Seconday Stratification grouping
- 'StratificationCategory3': Tertiary Stratification Category
- 'Stratification3': Tertiary stratification grouping 
- 'GeoLocation': Geolocation ID Point(x,y) 
- 'ResponseID': Nan
- 'LocationID', Numerical representation of Location (1-78)
- 'TopicID': Topic ID Abbreviation
- 'QuestionID': Question ID Abbreviation 
- 'DataValueTypeID': ID Abbreviation for Data Value Type
- 'StratificationCategoryID1: Stratificaiton Category ID 
- 'StratificationID1', Stratification ID
- 'StratificationCategoryID2': NAN
- 'StratificationID2': NAN
- 'StratificationCategoryID3': NAN
- 'StratificationID3': NAN

## Data Cleaning - Chronic Obstructive Pulmonary Disease (COPD)

```python
# Begin Disease Specified Cleaning
df = df_all.copy()
df.shape #should give (1185676, 34) - same as above
```

```python
df.isnull().sum() #Check for missing values from entire dataset
```

```python
# Drop Columns from entire dataset that are missing = Num of missing values == num of rows --> Completely empty
df = df.drop(columns=[
            'Response',  # Entire col is empty
            'StratificationCategory2', # Entire col is empty
            'Stratification2', # Entire col is empty
            'StratificationCategory3', # Entire col is empty
            'Stratification3', # Entire col is empty
            'ResponseID', # Entire col is empty
            'StratificationCategoryID2', # Entire col is empty
            'StratificationID2', # Entire col is empty
            'StratificationCategoryID3', # Entire col is empty
            'StratificationID3', # Entire col is empty
            'YearEnd', # We will use Year start instead
            'LocationAbbr', # We will use location instead
            'DataSource', #Not needed at this moment 
            'DataValueAlt', #Not needed, will use DataValue 
            'DataValueFootnoteSymbol', #Not needed at this moment 
            'DatavalueFootnote', #Not needed at this moment 
            'HighConfidenceLimit', #Not needed at this moment 
            'LowConfidenceLimit', #Not needed at this moment 
            'LocationID', # We will use Location
            'DataValueTypeID',  # Not needed 
            'TopicID' # we know this is copd
        ])
print(df.shape)
#should be (1185676, 14) - 20 cols were dropped
```

```python
# Now filter entier dataset for topic = 'Chronic Obstructive Pulmonary Disease'
topic = 'Chronic Obstructive Pulmonary Disease'
df_copd = df[df['Topic'] == topic]
print(df_copd.shape)
# df_copd.head()
# (1185676, 13) to (152874, 13) - about 12.9% of data is related to COPD
```

## Handling Missing values for Data Value

```python
# If data Value is missing, we have no option but to delete
# Total Rows: 152874, Total Missing Data Values: 58290
# percentage of missing values: 40%
# caused by : Data not shown because of too few respondents or cases (in the footnote)
df = df.dropna(subset=['DataValue']) 
df.isnull().sum()
```

```python
#Fill GeoLocation with Missing for now to avoid error
df['GeoLocation'] = df['GeoLocation'].fillna('Missing').astype(str)
```
